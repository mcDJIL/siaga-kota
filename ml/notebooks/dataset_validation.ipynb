# Dataset Validation

Notebook ini menampilkan validasi awal data banjir dan cuaca yang dipakai untuk pipeline ML produksi.

```python
import pandas as pd
from pathlib import Path

flood_path = Path('../datasets/raw/flood.csv')
weather_path = Path('../datasets/raw/weather_data.csv')

flood_df = pd.read_csv(flood_path)
weather_df = pd.read_csv(weather_path)

print('Flood shape:', flood_df.shape)
print('Weather shape:', weather_df.shape)
print('Flood columns:', list(flood_df.columns))
print('Weather columns:', list(weather_df.columns))
print('Flood missing values:', flood_df.isna().sum().to_dict())
print('Weather missing values:', weather_df.isna().sum().to_dict())
print('Flood duplicates:', flood_df.duplicated().sum())
print('Weather duplicates:', weather_df.duplicated().sum())

flood_df[['FloodProbability']].describe()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 5))
sns.histplot(flood_df['FloodProbability'], bins=30, kde=True)
plt.title('Distribusi Flood Probability')
plt.xlabel('Flood Probability')
plt.ylabel('Jumlah')
plt.tight_layout()
plt.show()
```
